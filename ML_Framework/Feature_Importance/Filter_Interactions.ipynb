```python
import pandas as pd

Nbins = 2
num_top = 25
folder = 'Feature_Importance'
property = 'Activity'
target_style = 'NG'
featurizer_style = 'All_Featurizers_' + target_style
ml_model_style = 'All_Models_' + target_style
importance_style = 'SHAP'
featurizer_name = 'RDKit_Descriptors_NG'
sheet_name = featurizer_name

# Define file paths.
importance_output_file = rf"C:\Users\grvkr\Box\Gaurav Kumar\Projects\SAR_NM\Codes\ML_Framework\Data_Combined\{folder}\{property}_{Nbins} bins_{importance_style}_Importance.xlsx"
interactions_output_file = rf"C:\Users\grvkr\Box\Gaurav Kumar\Projects\SAR_NM\Codes\ML_Framework\Data_Combined\{folder}\{property}_{Nbins} bins_{importance_style}_Interactions.xlsx"
output_file_top = rf"C:\Users\grvkr\Box\Gaurav Kumar\Projects\SAR_NM\Codes\ML_Framework\Data_Combined\{folder}\{property}_{Nbins} bins_{importance_style}_Top{num_top}_Interactions.xlsx"

# Load the importance data.
importance_df = pd.read_excel(importance_output_file, sheet_name=sheet_name)

# Sort by 'SHAP Importance' (highest first).
importance_df = importance_df.sort_values(by='SHAP Importance', ascending=False)

# Select the top num_top features based on SHAP Importance.
top_features = importance_df.head(num_top)['Feature Name'].tolist()

# Create a mapping from Feature Name to Feature ID.
feature_mapping = dict(zip(importance_df['Feature Name'], importance_df['Feature ID']))

# Load the interactions data.
interactions_df = pd.read_excel(interactions_output_file, sheet_name=sheet_name)

# Filter interactions so that both features are among the top features.
top_interactions = interactions_df[
    interactions_df['Feature 1'].isin(top_features) &
    interactions_df['Feature 2'].isin(top_features)
].copy()

# Add new columns for Feature 1 ID and Feature 2 ID.
top_interactions['Feature 1 ID'] = top_interactions['Feature 1'].map(feature_mapping)
top_interactions['Feature 2 ID'] = top_interactions['Feature 2'].map(feature_mapping)

# Rename feature columns to include "Name" in their header.
top_interactions = top_interactions.rename(columns={'Feature 1': 'Feature 1 Name', 'Feature 2': 'Feature 2 Name'})

# Reorder the columns.
new_order = ['Feature 1 ID', 'Feature 1 Name', 'Feature 2 ID', 'Feature 2 Name', 'Interaction Importance']
top_interactions = top_interactions[new_order]

# Save the filtered interactions to a new Excel file.
with pd.ExcelWriter(output_file_top, engine='openpyxl', mode='a', if_sheet_exists='replace') as writer:
    top_interactions.to_excel(writer, sheet_name=sheet_name, index=False)

print(f"Saved top {num_top} feature interactions to:", output_file_top)
```
