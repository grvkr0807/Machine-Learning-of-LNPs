# Save functions

```python

def save_class_info(datay_train, datay_test, classes, fold_idx, output_path):
    rows = []
    results_sheet_name= f'Fold {fold_idx + 1}'

    row = {}
    for cls in classes:
        row[f'Train_Count_{cls}'] = np.sum(datay_train == cls)
        row[f'Test_Count_{cls}'] = np.sum(datay_test == cls)
    rows.append(row)

    results_df = pd.DataFrame(rows)
    
    if not os.path.exists(output_path):
        with pd.ExcelWriter(output_path, engine='openpyxl', mode='w') as writer:
            results_df.to_excel(writer, sheet_name=results_sheet_name, index=False)
    else:
        with pd.ExcelWriter(output_path, engine='openpyxl', mode='a', if_sheet_exists='replace') as writer:
            results_df.to_excel(writer, sheet_name=results_sheet_name, index=False)



```
