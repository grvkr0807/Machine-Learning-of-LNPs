```python
import pandas as pd
from pandas.api.types import CategoricalDtype

# File paths and other parameters
Nbins= 2
folder= '5Fold_CV'
property= 'Cell_Viability'
lipid_num=1
input_file = rf"C:\Users\grvkr\Box\Gaurav Kumar\Projects\SAR_NM\Codes\ML_Framework\Data_Combined\{folder}\{property}_Metrics {Nbins} bins - Copy.xlsx"
output_file = rf"C:\Users\grvkr\Box\Gaurav Kumar\Projects\SAR_NM\Codes\ML_Framework\Data_Combined\{folder}\{property}_Metrics {Nbins} bins.xlsx"

# Define the desired orders
ml_model_order = [
    "ML_Model_RF", 
    "ML_Model_ExtraTrees", 
    "ML_Model_GB",
    "ML_Model_SVM_Linear", 
    "ML_Model_GCN",
    "ML_Model_GAT",
    "ML_Model_AdaB",
    "ML_Model_SVM_RBF"
    
]

featurizer_order = [
    "RDKit_Descriptors_NG", 
    "Descriptastorus_RDKit2DNormalized_NG",
    "RDKit_3DDescriptors_NG",
    "RDKit_MACCS_NG",
    "RDKit_FPGenerator_NG", 
    "RDKit_AtomPairGenerator_NG", 
    "RDKit_MorganGenerator_NG", 
    "PytorchGeometric_NG", 
    "DeepChem_MolGraphConv_NG", 
    "DeepChem_DMPNN_NG",
    "Hybrid_Descriptors_Graph_NG", 
    "Hybrid_Fingerprints_Graph_NG", 
    
    "PytorchGeometric_GG",
    "DeepChem_MolGraphConv_GG",
    "DeepChem_DMPNN_GG",
    "Hybrid_Descriptors_Graph_GG", 
    "Hybrid_Fingerprints_Graph_GG"
]

# Create categorical types with the specified order
ml_cat_type = CategoricalDtype(categories=ml_model_order, ordered=True)
feat_cat_type = CategoricalDtype(categories=featurizer_order, ordered=True)


# Read the Excel file
with pd.ExcelFile(input_file) as xls:
    # We'll only sort sheets that start with "Fold " (i.e. Fold 1, Fold 2, etc.)
    fold_sheet_names = [sheet for sheet in xls.sheet_names if sheet.startswith("Fold ")]
    
    # Dictionary to hold sorted DataFrames per sheet
    sorted_sheets = {}
    
    for sheet in fold_sheet_names:
        df = pd.read_excel(xls, sheet_name=sheet)
        
        # Convert 'ML_Model' and 'Featurizer' columns to categorical with desired order
        df["ML_Model"] = df["ML_Model"].astype(ml_cat_type)
        df["Featurizer"] = df["Featurizer"].astype(feat_cat_type)
        
        # Sort by ML_Model first and then Featurizer
        df_sorted = df.sort_values(by=["ML_Model", "Featurizer"])
        sorted_sheets[sheet] = df_sorted

# Write the sorted sheets to a new Excel file
with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    for sheet, sorted_df in sorted_sheets.items():
        sorted_df.to_excel(writer, sheet_name=sheet, index=False)

print("Sheets have been sorted and saved to:", output_file)

```
